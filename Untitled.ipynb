```python
class Platform:
    def __init__(self,name,query_format=None):
        self.name = name
        self.query_format = query_format
        self.html = ''

    def take_format(self, query_format):
        self.query_format = query_format

    def get_query_structure(self):
        return self.query_format
    
    def get_query(self,prompt):
        query_format = self.query_format
        query = query_format+prompt
        return query

    def make_list_cards(self) -> list:
        raise NotImplementedError

    def get_info_card(self, card):
        raise NotImplementedError

    def get_results(self):
        cards = self.make_list_cards()
        results = []
        for card in cards:
            results.append(self.get_info_card(card))
        return results
```

```python
class Croma(Platform):
    def __init__(self):
        super().__init__("Croma", "https://www.croma.com/searchB?q=")
    
    def get_query(self,prompt):
        query_format = self.query_format
        query = "https://www.croma.com/searchB?q={prompt}%3Arelevance&text={prompt}"
        return query
```

```python
import re
from bs4 import BeautifulSoup
from base_platform import Platform
import asyncio
from playwright.async_api import async_playwright 


class Croma(Platform):
    def __init__(self):
        super().__init__("Croma")
    
    def get_query(self,prompt):
        encoded = prompt.replace(" ", "%20")
        self.query = url = f"https://www.croma.com/searchB?q={encoded}%3Arelevance&text={encoded}"
        return self.query
        
    def make_list_cards(self) -> list:
            soup = BeautifulSoup(self.html, 'html.parser')
            valid_cards = []
            for card in soup.find_all("li", class_="product-item"):
                if card.find("h3", class_="product-title"):
                    valid_cards.append(card)
            return valid_cards

    def get_info_card(self, card):
        try:
            h3_elem = card.find("h3", class_="product-title")
            title = h3_elem.text.strip() if h3_elem else "N/A"

            link = card.find("a", href=True)
            if link and link["href"]:
                href = link["href"]
                url = href if href.startswith("http") else f"https://www.croma.com{href}"
            else:
                url = "N/A"

            amount_elem = card.find("span", class_="amount")
            if amount_elem:
                clean_digits = re.sub(r"[^\d]", "", amount_elem.text)
                selling_price = int(clean_digits) if clean_digits else 0
            else:
                selling_price = 0

            img_elem = card.find("img")
            image = ""
            if img_elem:
                image = img_elem.get("src") or img_elem.get("data-src") or ""

            return {
                'platform': 'Croma',
                'title': title,
                'brand': title.split(' ')[0] if title != "N/A" else "N/A",
                'price': selling_price,
                'url': url,
                'image': image
            }
        except Exception:
                return None
```
